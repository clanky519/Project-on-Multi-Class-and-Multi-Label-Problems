```python
from sklearn.svm import LinearSVC, SVC
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.multiclass import OneVsRestClassifier, OneVsOneClassifier, OutputCodeClassifier
from sklearn.model_selection import train_test_split
from sklearn.model_selection import KFold, cross_validate
```

```python
from sklearn.base import BaseEstimator, ClassifierMixin
from sklearn.tree import DecisionTreeClassifier
import numpy as np

class DAGClassifier(BaseEstimator, ClassifierMixin):
    def init(self):
        self.classifiers = {}
        self.classes = None

    def fit(self, X, y):
        # Convert to numpy (FIX)
        X = np.array(X)
        y = np.array(y)

        self.classes = np.unique(y)
        self.classifiers = {}

        for i in range(len(self.classes)):
            for j in range(i+1, len(self.classes)):
                c1, c2 = self.classes[i], self.classes[j]

                idx = np.where((y == c1) | (y == c2))[0]  # FIX: [0]
                X_pair, y_pair = X[idx], y[idx]

                clf = DecisionTreeClassifier(random_state=42)
                clf.fit(X_pair, y_pair)

                self.classifiers[(c1, c2)] = clf

        return self

    def predict(self, X):
        X = np.array(X)  # FIX
        predictions = []

        for x in X:
            classes = list(self.classes)

            while len(classes) > 1:
                c1, c2 = classes[0], classes[1]
                clf = self.classifiers[(min(c1,c2), max(c1,c2))]

                pred = clf.predict([x])[0]

                if pred == c1:
                    classes.remove(c2)
                else:
                    classes.remove(c1)

            predictions.append(classes[0])

        return np.array(predictions)
```

```python
#base classifiers
lr = LogisticRegression(max_iter=1000)
svm = LinearSVC(max_iter=1000)
dt = DecisionTreeClassifier(criterion='entropy')
bc = [lr, svm, dt]
ovr_lr = OneVsRestClassifier(lr)
ovr_svm = OneVsRestClassifier(svm)
ovr_dt = OneVsRestClassifier(dt)
ovo_lr = OneVsOneClassifier(lr)
ovo_svm = OneVsOneClassifier(svm)
ovo_dt = OneVsOneClassifier(dt)
ecoc_lr = OutputCodeClassifier(lr)
ecoc_svm = OutputCodeClassifier(svm)
ecoc_dt = OutputCodeClassifier(dt)
mdt = DecisionTreeClassifier(criterion='entropy')
msvm = SVC(decision_function_shape='ovo')
dag = DAGClassifier()
#multiclass classifiers
models = [ovr_lr, ovr_svm, ovr_dt, ovo_lr, ovo_svm, ovo_dt, ecoc_lr, ecoc_svm, ecoc_dt, mdt, msvm, dag]
kf = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
scoring = {
    'accuracy': 'accuracy',
    'f1_macro': 'f1_macro',
    'f1_micro': 'f1_micro',
    'precision_macro': 'precision_macro',
    'precision_micro': 'precision_micro',
    'recall_macro': 'recall_macro',
    'recall_micro': 'recall_micro'
}
```

```python
#df1
y_blood = df['RSRP']
X_blood = df.drop(columns=['RSRP'])
Xtr_blood, Xt_blood, ytr_blood, yt_blood = train_test_split(X_blood, y_blood, test_size=0.2, random_state=42)
```

```python
scores = []
for model in models:
  score = cross_validate(model, X_blood, y_blood, cv=kf, scoring=scoring)
  auc_mean = round(score['test_accuracy'].mean(), 2)
  auc_std = round(score['test_accuracy'].std(), 2)
  recall_ma_mean = round(score['test_recall_macro'].mean(), 2)
  recall_ma_std = round(score['test_recall_macro'].std(), 2)
  recall_mi_mean = round(score['test_recall_micro'].mean(), 2)
  recall_mi_std = round(score['test_recall_micro'].std(), 2)
  precision_ma_mean = round(score['test_precision_macro'].mean(), 2)
  precision_ma_std = round(score['test_precision_macro'].std(), 2)
  precision_mi_mean = round(score['test_precision_micro'].mean(), 2)
  precision_mi_std = round(score['test_precision_micro'].std(), 2)
  f1_ma_mean = round(score['test_f1_macro'].mean(), 2)
  f1_ma_std = round(score['test_f1_macro'].std(), 2)
  f1_mi_mean = round(score['test_f1_macro'].mean(), 2)
  f1_mi_std = round(score['test_f1_macro'].std(), 2)
  scores.append([[auc_mean, auc_std], [recall_ma_mean, recall_ma_std], [recall_mi_mean, recall_mi_std], [precision_ma_mean, precision_ma_std], [precision_mi_mean, precision_mi_std], [f1_ma_mean, f1_ma_std], [f1_mi_mean, f1_mi_std]])
  print("DONE")
```

```python
col = ["OVR_lr", "OVR_svm", "OVR_dt", "OVO_lr", 'OVO_svm', 'OVO_dt', "ECOC_lr", 'ECOC_svm', 'ECOC_dt', 'Multiclass DT', 'Multiclass SVM', 'DAG']
scs = []
for i in range(len(scores)):
  sc = [col[i]]
  for s in scores[i]:
    t = str(float(s[0]))+"+"+str(float(s[1]))
    sc.append(t)
  scs.append(sc)
cols = ['Model', "Micro Accuracy", "Macro Recall", "Micro Recall", "Macro Precision", "Micro Precision", "Macro F1", "Micro F1"]
table = pd.DataFrame(scs, columns=cols)
table
```
